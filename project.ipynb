```python
import numpy #to use with image arrays
import cv2 
import matplotlib.pyplot as plt
```

```python
#Create a function that takes an integer and a boolean as inputs and hides the boolean value in the last
#significant bit of the integer
        
def hide_boolean_into_integer(integer, boolean):
    
        if boolean == True:
            bln = 1
        elif boolean == False:
            bln = 0
        else:
            return None
        if integer % 2==bln:
            return integer
        elif integer % 2 != bln:
            if bln == 1:
                integer = integer + 1
            else:
                integer = integer - 1
        return integer
```

```python
#function that retriev the last significant bit
def retrieval_bolean_from_integer(retrive):
    if retrive % 2 == 1:
        return 1
    else:
        return 0
```

```python
def hide_binary_in_grayscale(g,b):
    g[:,:] &=0b11111110 #mask the the unused bits and put the last bit to zero
    g[:,:] |= b   #hide in the last bit
    return g
```

```python
# retrieve a binary image from a grayscale image

def retrieve_binary_from_grayscale(grayscale):
    grayscale = grayscale.copy() 
    rows, columns=grayscale.shape
        
    binary=numpy.zeros((rows, columns), dtype=numpy.uint8) #generate 3 binary images , all the pixels are 0
        

    grayscale[:,:] &=0b00000001

    binary= grayscale[:,:] | binary
        

    return binary
```

```python
####################### Hide three different binary images in an RGB image#######################################

def threebinary_in_RGB(RGBimage,binary_image1,binary_image2,binary_image3):
    RGBimage = RGBimage.copy()   #make a copy
    if RGBimage.shape[2] !=3:    #check if it is RGB or not
        print('it is not RGB image')
    else:
        rows, columns,channels=RGBimage.shape # extract rows, columns,channels
        
        if binary_image1.shape == (rows,columns) and binary_image2.shape == (rows,columns) and binary_image3.shape == (rows,columns): #check the size

             RGBimage[:,:,0] &=0b11111110 #mask the the unused bits and put the last bit to zero
             RGBimage[:,:,1] &=0b11111110 #mask the the unused bits and put the last bit to zero
             RGBimage[:,:,2] &=0b11111110 #mask the the unused bits and put the last bit to zero
             
             RGBimage[:,:,0] |= binary_image1   #hide in the last bit
             RGBimage[:,:,1] |= binary_image2
             RGBimage[:,:,2] |= binary_image3
             return RGBimage
        else:
            print ('the images are not the same size')
```

```python
#retrieve three different binary images in an RGB image

def retrieve_3binary_from_RGB(R):
    R = R.copy() 
    if R.shape[2] !=3:
        print('it is not RGB image')
    else:
        rows, columns,channels=R.shape
        
        binary1=numpy.zeros((rows, columns), dtype=numpy.uint8) #generate 3 binary images , all the pixels are 0
        binary2=numpy.zeros((rows, columns), dtype=numpy.uint8)
        binary3=numpy.zeros((rows, columns), dtype=numpy.uint8)

        R[:,:,0] &=0b00000001
        R[:,:,1] &=0b00000001
        R[:,:,2] &=0b00000001

        binary1=R[:,:,0] | binary1
        binary2=R[:,:,1] | binary2
        binary3=R[:,:,2] | binary3
        
        plt.figure(figsize=(10,10))
        plt.subplot(1,3,1)
        plt.imshow(binary1,cmap='gray')
        plt.subplot(1,3,2)
        plt.imshow(binary2,cmap='gray')
        plt.subplot(1,3,3)
        plt.imshow(binary3,cmap='gray')
        plt.axis('off')
        plt.show()

    return 

```

```python
#Hide a grayscale image in a RGB source image by storing two bits of data in the red channel and three in the green and blue ones.

def grayscale_into_RGB (image1,image2):
    image1= image1.copy()
    if image1.shape[2] !=3:
        print('it is not RGB image')
    else:
        rows, columns,channels=image1.shape
        if image2.shape == (rows,columns):
            rows2= image2.shape[0]
            columns2= image2.shape[1]
            bits7_8=numpy.zeros((rows2, columns2), dtype=numpy.uint8)
            bits6_5_4=numpy.zeros((rows2, columns2), dtype=numpy.uint8)
            bits3_2_1=numpy.zeros((rows2, columns2), dtype=numpy.uint8)
            for i in range (rows2):
                for j in range (columns2):
                    
                    bits7_8[i,j]=(image2[i,j] >> 6)
                    bits6_5_4[i,j]=((image2[i,j] >> 3) & 0b111)
                    bits3_2_1[i,j]=(image2[i,j] & 0b111)

                    image1[i,j,0]&=0b11111100
                    image1[i,j,1]&=0b11111000
                    image1[i,j,2]&=0b11111000

                    image1[i,j,0]|= bits7_8[i,j]
                    image1[i,j,1]|= bits6_5_4[i,j]
                    image1[i,j,2]|= bits3_2_1[i,j]

    return image1
```

```python
#Extract and reconstruct the hidden grayscale image.

def retrieve_grayscale_from_RGB(R1):
    R1 = R1.copy() 
    if R1.shape[2] !=3:
        print('it is not RGB image')
    else:
        rows, columns,channels=R1.shape

        grayscale=numpy.zeros((rows, columns), dtype=numpy.uint8)

        R1[:,:,0] &=0b00000011
        R1[:,:,1] &=0b00000111
        R1[:,:,2] &=0b00000111

        grayscale= ((R1[:,:,0] | grayscale) << 6) | ((R1[:,:,1] | grayscale) << 3) | (R1[:,:,2] | grayscale)
    return grayscale
        
```

```python
def hide_multiple_bits_of_data_per_pixel(im,n,data):
    i=[254,252,248,240,224,192,128,0]
        
    im[:,:] &=i[n-1] #mask the the unused bits and put the last bit to zero
    im[:,:] |= data
    return im
```
